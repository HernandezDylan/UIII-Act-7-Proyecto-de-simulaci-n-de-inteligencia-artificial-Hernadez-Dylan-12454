```python
print("====================================")
print("SISTEMA INTELIGENTE PARA ANALISTA")
print("====================================")

while True:

    print("\n=========== MENU ===========")
    print("1 Analizar cliente")
    print("2 Verificar seguridad")
    print("3 Consultar datos")
    print("4 Estado del sistema")
    print("5 Salir")

    opcion = input("Selecciona opcion: ")


    # OPCION 1
    if opcion == "1":

        print("\n====== ANALISIS DEL CLIENTE ======")

        cliente = input("Nombre del cliente: ")

        empresa = input("Empresa: ")

        datos = int(input("Cantidad de datos analizados: "))

        riesgo = int(input("Nivel de riesgo del 1 al 10: "))


        if riesgo >= 8:

            print("RIESGO MUY ALTO")

        elif riesgo >= 5:

            print("RIESGO MEDIO")

        else:

            print("RIESGO BAJO")


        if datos >= 1000:

            print("MUCHOS DATOS ANALIZADOS")

        elif datos >= 500:

            print("DATOS MODERADOS")

        else:

            print("POCOS DATOS")


        print("Cliente:", cliente)

        print("Empresa:", empresa)


    # OPCION 2
    elif opcion == "2":

        print("\n====== VERIFICACION DE SEGURIDAD ======")

        seguridad = int(input("Nivel de seguridad: "))

        intentos = int(input("Intentos incorrectos: "))


        if seguridad >= 8 and intentos <= 2:

            print("SEGURIDAD ALTA")

        elif seguridad >= 5:

            print("SEGURIDAD MEDIA")

        else:

            print("SEGURIDAD BAJA")


    # OPCION 3
    elif opcion == "3":

        print("\n====== CONSULTA DE DATOS ======")

        usuarios = int(input("Cantidad de usuarios: "))

        servidores = int(input("Cantidad de servidores: "))


        if usuarios >= 100:

            print("MUCHOS USUARIOS")

        else:

            print("POCOS USUARIOS")


        if servidores >= 10:

            print("SISTEMA GRANDE")

        else:

            print("SISTEMA PEQUEÑO")


    # OPCION 4
    elif opcion == "4":

        print("\n====== ESTADO DEL SISTEMA ======")

        alertas = int(input("Cantidad de alertas: "))

        fallas = int(input("Cantidad de fallas: "))


        if alertas >= 10 or fallas >= 5:

            print("SISTEMA EN RIESGO")

        elif alertas >= 5:

            print("SISTEMA CON SUPERVISION")

        else:

            print("SISTEMA ESTABLE")


    # OPCION 5
    elif opcion == "5":

        print("Programa finalizado")

        break


    # OPCION INCORRECTA
    else:

        print("Opcion incorrecta")
```
